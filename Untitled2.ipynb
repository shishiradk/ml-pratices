```python
# Install required packages
!pip install openai pytesseract pillow PyPDF2
!apt-get install tesseract-ocr -y

# Import libraries
import openai
import pytesseract
import PyPDF2
from PIL import Image
from google.colab import files
import getpass

# Securely enter your OpenAI API key
openai.api_key = getpass.getpass("🔑 Enter your OpenAI API key:")

# Function to extract text from uploaded files
def extract_text_from_uploaded_file(uploaded_file, file_type):
    if file_type == 'pdf':
        reader = PyPDF2.PdfReader(uploaded_file)
        text = ''
        for page in reader.pages:
            text += page.extract_text()
        return text
    elif file_type == 'img':
        image = Image.open(uploaded_file)
        return pytesseract.image_to_string(image)
    elif file_type == 'txt':
        text = uploaded_file.read().decode('utf-8')
        return text
    else:
        raise ValueError("Unsupported file type.")

# Upload syllabus file
print("📄 Upload syllabus file (PDF, Image, or TXT):")
syllabus_uploaded = files.upload()

syllabus_filename = list(syllabus_uploaded.keys())[0]
syllabus_filetype = syllabus_filename.split('.')[-1].lower()

if syllabus_filetype in ['jpg', 'jpeg', 'png']:
    syllabus_text = extract_text_from_uploaded_file(syllabus_filename, 'img')
elif syllabus_filetype == 'pdf':
    syllabus_text = extract_text_from_uploaded_file(syllabus_filename, 'pdf')
elif syllabus_filetype == 'txt':
    syllabus_text = extract_text_from_uploaded_file(open(syllabus_filename, 'rb'), 'txt')
else:
    raise ValueError("Unsupported syllabus file type.")

# Upload past questions file
print("📄 Upload past questions file (PDF, Image, or TXT):")
past_q_uploaded = files.upload()

past_q_filename = list(past_q_uploaded.keys())[0]
past_q_filetype = past_q_filename.split('.')[-1].lower()

if past_q_filetype in ['jpg', 'jpeg', 'png']:
    past_q_text = extract_text_from_uploaded_file(past_q_filename, 'img')
elif past_q_filetype == 'pdf':
    past_q_text = extract_text_from_uploaded_file(past_q_filename, 'pdf')
elif past_q_filetype == 'txt':
    past_q_text = extract_text_from_uploaded_file(open(past_q_filename, 'rb'), 'txt')
else:
    raise ValueError("Unsupported past questions file type.")

# Create prompt
prompt = (
    f"Here is the syllabus:\n{syllabus_text}\n\n"
    f"Here are past model questions:\n{past_q_text}\n\n"
    "Generate 10 new model questions following the same style."
)

# Call GPT-3.5 model using OpenAI's new API
try:
    response = openai.chat.completions.create(
        model="gpt-3.5-turbo",  # ✅ GPT-3.5 is accessible to everyone
        messages=[{"role": "user", "content": prompt}]
    )

    print("\n🧠 Generated Model Questions:\n")
    print(response.choices[0].message.content)

except Exception as e:
    print(f"⚠️ OpenAI API Error: {e}")
```
