```python
# ✅ 2. Import Libraries
from transformers import pipeline, AutoModelForSeq2SeqLM, AutoTokenizer
from keybert import KeyBERT
from sentence_transformers import SentenceTransformer, util
import torch
from google.colab import files
```

```python
# ✅ 3. Load Models
model_name = "google/flan-t5-small"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)
qg_pipeline = pipeline("text2text-generation", model=model, tokenizer=tokenizer, max_length=128)
# ✅ 3. Load Models
model_name = "google/flan-t5-small"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)
qg_pipeline = pipeline("text2text-generation", model=model, tokenizer=tokenizer, max_length=128)

embedding_model = SentenceTransformer('all-MiniLM-L6-v2')
kw_model = KeyBERT(model=embedding_model)
```

```python
# ✅ 4. Define Functions

def extract_topics(syllabus_text, num_topics=5):
    keywords = kw_model.extract_keywords(syllabus_text, keyphrase_ngram_range=(1, 2), stop_words='english', top_n=num_topics)
    topics = [kw[0] for kw in keywords]
    return topics

def generate_questions(syllabus, example_questions, num_questions=5):
    prompt = f"""
    Syllabus: {syllabus}
    Example Questions: {example_questions}
    Task: Generate {num_questions} new model questions based on the syllabus following the style of the example questions.
    """
    outputs = qg_pipeline(prompt)
    questions = [output['generated_text'] for output in outputs]
    return questions

def match_to_syllabus(questions, topics, threshold=0.5):
    filtered_questions = []
    topic_embeddings = embedding_model.encode(topics, convert_to_tensor=True)

    for question in questions:
        question_embedding = embedding_model.encode(question, convert_to_tensor=True)
        cosine_scores = util.cos_sim(question_embedding, topic_embeddings)
        max_score = torch.max(cosine_scores).item()

        if max_score >= threshold:
            filtered_questions.append(question)
    return filtered_questions

def generate_model_questions(syllabus_text, example_questions_text, num_questions=5):
    print("Extracting topics from syllabus...")
    topics = extract_topics(syllabus_text)
    print(f"Extracted Topics: {topics}")

    print("\nGenerating new questions...")
    generated_questions = generate_questions(syllabus_text, example_questions_text, num_questions*2)  # Generate extra for filtering
    print(f"\nGenerated Raw Questions: {generated_questions}")

    print("\nFiltering questions based on syllabus...")
    final_questions = match_to_syllabus(generated_questions, topics)

    return final_questions[:num_questions]  # Limit to requested number
```

```python
import fitz  # PyMuPDF
from fpdf import FPDF
from ipywidgets import FileUpload, Textarea, Button, VBox, HBox, Label
from IPython.display import display
from PIL import Image
import pytesseract
import io
import os

# Initialize storage
uploaded_syllabus_images = []
uploaded_model_question_images = []
model_questions_text = ""

# 1. Helper Functions
def extract_pdf_text(pdf_bytes):
    text = ""
    doc = fitz.open(stream=pdf_bytes, filetype="pdf")
    for page in doc:
        text += page.get_text()
    return text

# PDF class with custom font
class PDF(FPDF):
    def __init__(self):
        super().__init__()
        # Add a custom font (no need for `uni=True` in newer FPDF versions)
        self.add_font('DejaVu', '', 'path_to_your_font/DejaVuSans.ttf')

def create_pdf(text, filename):
    pdf = PDF()
    pdf.set_auto_page_break(auto=True, margin=15)
    pdf.add_page()
    pdf.set_font("DejaVu", size=12)
    pdf.multi_cell(0, 10, text)
    pdf.output(filename)

def create_pdf_with_images(text, images, filename):
    pdf = PDF()
    pdf.set_auto_page_break(auto=True, margin=15)
    pdf.add_page()
    pdf.set_font("DejaVu", size=12)
    pdf.multi_cell(0, 10, text)

    for img in images:
        pdf.add_page()
        img_byte_arr = io.BytesIO()
        img.save(img_byte_arr, format='JPEG')
        img_byte_arr.seek(0)
        with open('temp_img.jpg', 'wb') as f:
            f.write(img_byte_arr.read())
        pdf.image('temp_img.jpg', x=10, y=30, w=pdf.w - 20)

    pdf.output(filename)
    # Clean up temporary image
    if os.path.exists('temp_img.jpg'):
        os.remove('temp_img.jpg')

# 2. Upload Widgets
upload_syllabus = FileUpload(accept='.pdf,.jpg,.jpeg,.png', multiple=True)
upload_model_questions = FileUpload(accept='.pdf,.jpg,.jpeg,.png', multiple=True)

# 3. Text Inputs
syllabus_input = Textarea(
    placeholder='Syllabus content will appear here...',
    description='Syllabus:',
    layout={'width': '90%', 'height': '150px'}
)

model_questions_input = Textarea(
    placeholder='Model questions will appear here...',
    description='Model Questions:',
    layout={'width': '90%', 'height': '150px'}
)

# 4. Status Labels
status_label = Label()
status_label2 = Label()

# 5. Handle Uploads
def handle_syllabus_upload(change):
    global uploaded_syllabus_images
    status_label.value = "⏳ Extracting syllabus..."
    files = upload_syllabus.value

    for name, fileinfo in files.items():
        if name.endswith('.pdf'):
            extracted_text = extract_pdf_text(fileinfo['content'])
            syllabus_input.value += "\n" + extracted_text
        elif name.endswith(('.jpg', '.jpeg', '.png')):
            img = Image.open(io.BytesIO(fileinfo['content']))
            uploaded_syllabus_images.append(img)

    status_label.value = "✅ Syllabus uploaded!"

def handle_model_questions_upload(change):
    global uploaded_model_question_images, model_questions_text
    status_label2.value = "⏳ Extracting model questions..."
    files = upload_model_questions.value

    for name, fileinfo in files.items():
        if name.endswith('.pdf'):
            extracted_text = extract_pdf_text(fileinfo['content'])
            model_questions_input.value += "\n" + extracted_text
        elif name.endswith(('.jpg', '.jpeg', '.png')):
            img = Image.open(io.BytesIO(fileinfo['content']))
            uploaded_model_question_images.append(img)
            # OCR: Extract text from image
            extracted_text = pytesseract.image_to_string(img)
            model_questions_input.value += "\n" + extracted_text

    status_label2.value = "✅ Model questions uploaded!"

upload_syllabus.observe(handle_syllabus_upload, names='value')
upload_model_questions.observe(handle_model_questions_upload, names='value')

# 6. Add Manually Model Question
def add_model_question(change):
    global model_questions_text
    model_questions_text += model_questions_input.value + "\n"
    model_questions_input.value = ""

add_question_button = Button(description="➕ Add Model Question", button_style='info')
add_question_button.on_click(add_model_question)

# 7. Done Button
def create_all_pdfs(b):
    if not syllabus_input.value.strip():
        print("⚠️ Please upload or type the syllabus.")
        return
    if not model_questions_input.value.strip():
        print("⚠️ Please upload or type model questions.")
        return

    create_pdf_with_images(syllabus_input.value, uploaded_syllabus_images, "syllabus.pdf")
    create_pdf(model_questions_text, "model_questions.pdf")
    print("✅ All PDFs created successfully!")

done_button = Button(description="🎯 Finish and Create PDFs", button_style='success')
done_button.on_click(create_all_pdfs)

# 8. Display
display(VBox([
    Label("📂 Upload Syllabus Files (.pdf, .jpg, .png):"),
    upload_syllabus,
    status_label,
    syllabus_input,

    Label("📂 Upload Model Questions Files (.pdf, .jpg, .png):"),
    upload_model_questions,
    status_label2,
    model_questions_input,
    add_question_button,

    done_button
]))
```
