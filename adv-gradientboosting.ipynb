Import libraries

Load the dataset

```python
#load iris dataset
data=load_iris()

#create a Dataframe
X= pd.DataFrame(data.data,columns=data.feature_names)
Y= pd.DataFrame(data.target,columns=['target'])

#Display the first rows of the dataset
print(X.head())
print(Y.head())
```
